```python
import numpy as np
from roi_selector import ROISelector, crop_stack
```

```python
def generate_sample_stack(n, height, width):
    # Create 2D coordinate grid
    x = np.linspace(-1, 1, width)
    y = np.linspace(-1, 1, height)
    X, Y = np.meshgrid(x, y)

    # Generate 2D Gaussian base image
    sigma_x = 0.3
    sigma_y = 0.3

    base_image = np.exp(
        -(X**2 / (2 * sigma_x**2) + Y**2 / (2 * sigma_y**2))
    ).astype(np.float32)

    # Generate independent Gaussian noise
    noise = np.random.normal(
        loc=0,
        scale=0.05,
        size=(n, height, width)
    ).astype(np.float32)

    # Generate image stack
    stack = np.clip(base_image[np.newaxis, :, :] + noise, 0, 1)

    return stack
```

```python
datr = generate_sample_stack(10, 2024, 2024)
datr.shape
```

```python
selector = ROISelector(stack=datr, display_index=5)
cropped_stack = crop_stack(datr, selector.selected_roi)
```
